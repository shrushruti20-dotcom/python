```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

data = pd.read_csv('Titanic Dataset (1).csv')

data.head(5)

```

```python

sns.boxplot(data = data, x = 'Embarked', y = 'Age')
plt.show()
```

```python

plt.scatter(x = data['Fare'], y = data['Survived'])
plt.ylabel('Survived')
plt.xlabel('Fare')
plt.show()
```

```python

plt.scatter(x = data['Parch'], y = data['Survived'])
plt.ylabel('Survived')
plt.xlabel('Parch')
plt.show()
```
